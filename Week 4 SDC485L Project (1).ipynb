```python
from IPython.display import display, Markdown
import datetime
import pandas as pd
from sklearn.preprocessing import StandardScaler

student_id = "JOSZIM1500"
todays_date = datetime.date.today().strftime("%B %d, %Y")

display(Markdown("### Step 4.0: Data Preparation and Feature Scaling"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Load dataset
df = pd.read_csv("water_potability.csv")

# Handle missing values
df.fillna(df.mean(), inplace=True)

# Separate features and target
X = df.drop("Potability", axis=1)
y = df["Potability"]

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Feature matrix shape:", X_scaled.shape)
print("Target vector shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

display(Markdown("### Step 4.1: Classification Model Optimization"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

param_grid = {
    "C": [0.01, 0.1, 1, 10],
    "penalty": ["l2"],
    "solver": ["lbfgs"]
}

grid_clf = GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid,
    cv=5
)
grid_clf.fit(X_train, y_train)

# Retrain with best parameters
best_clf = grid_clf.best_estimator_
best_clf.fit(X_train, y_train)

# Evaluate
y_pred = best_clf.predict(X_test)
clf_accuracy = accuracy_score(y_test, y_pred)
clf_loss = 1 - clf_accuracy

print("Classification Accuracy:", clf_accuracy)
print("Classification Loss:", clf_loss)
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

display(Markdown("### Step 4.2: Regression Model Optimization"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

y_reg = df["Hardness"]

Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X_scaled, y_reg, test_size=0.2, random_state=42
)

ridge_params = {"alpha": [0.1, 1, 10, 50]}

grid_ridge = GridSearchCV(Ridge(), ridge_params, cv=5)
grid_ridge.fit(Xr_train, yr_train)

# Retrain with best parameters
best_ridge = grid_ridge.best_estimator_
best_ridge.fit(Xr_train, yr_train)

# Evaluate
yr_pred = best_ridge.predict(Xr_test)
reg_mse = mean_squared_error(yr_test, yr_pred)

print("Regression Mean Squared Error:", reg_mse)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

display(Markdown("### Step 4.3: Clustering Model Optimization"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

silhouette_scores = {}

# Hyperparameter tuning for k
for k in range(2, 6):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    silhouette_scores[k] = silhouette_score(X_scaled, labels)

best_k = max(silhouette_scores, key=silhouette_scores.get)

# Retrain final model
final_kmeans = KMeans(n_clusters=best_k, random_state=42)
cluster_labels = final_kmeans.fit_predict(X_scaled)

cluster_score = silhouette_score(X_scaled, cluster_labels)

print("Optimal number of clusters (k):", best_k)
print("Silhouette Score:", cluster_score)
```
